# Préparation des données

```python
import pandas as pd
df = pd.read_csv("../data/thelook_fr_women_2023_2024.csv")
```

## Conversion des dates

```python
col_date=["item_created_at", "order_created_at", "shipped_at", "delivered_at"]
df[col_date] = df[col_date].apply(pd.to_datetime, errors='coerce')

df.info()
```

## Valeurs manquantes et doublons

```python
df.isna().sum()

# remplacement des valeurs manquantes par "Unknown" pour la colonne "brand"
df["brand"]=df["brand"].fillna("Unknown") 

df.duplicated().sum()
```

## Cohérence des bornes temporelles

```python
mask = (
    (df["order_created_at"] > df["shipped_at"]) |
    (df["shipped_at"] > df["delivered_at"])
)

df_incoherent = df[mask]

print("Nombre de commandes incohérentes :", len(df_incoherent))
```
