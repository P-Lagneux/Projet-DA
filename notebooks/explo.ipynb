## Description des données

donnés d'un site de vente de vêtement en ligne.
Le dataset est composé de 7 tables principales :

- users : Informations démographiques et localisation des clients.
- orders : Statut, date de création et détails globaux des commandes.
- order_items : Les articles spécifiques commandés pour chaque panier (prix de vente, retours, etc.).
- products : Catalogue des vêtements (marques, catégories, prix de vente conseillé).
- inventory_items : État des stocks et coûts d'approvisionnement des produits.
- distribution_centers : Données sur les 10 entrepôts logistiques (situés aux États-Unis).
- events : Données de navigation web (clics, sessions, parcours d'achat).
