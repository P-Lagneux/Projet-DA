# Exploration descriptive et visualisation

```python
import pandas as pd
import seaborn as sns
```

## Statistiques descriptives

```python
df = pd.read_parquet("df.parquet")

# totals des ventes et des couts

total_ventes = df['sale_price'].sum()
total_couts = df['cost'].sum()
marge_totale = total_ventes - total_couts

print(f"Avec un total de ventes de {total_ventes:,.2f} € et un total de coûts de {total_couts:,.2f} €, la marge totale est de {marge_totale:,.2f} €.")

# moyennes totales 

moyenne_ventes = df['sale_price'].mean()
moyenne_couts = df['cost'].mean()
marge_moyenne = moyenne_ventes - moyenne_couts

print(f"La moyenne des ventes est de {moyenne_ventes:,.2f} € pour un coût moyen de {moyenne_couts:,.2f} €. La marge moyenne est donc de {marge_moyenne:,.2f} €.")

# statistiques globales

stats_ventes= df['sale_price'].describe()
stats_couts= df['cost'].describe()
stats = pd.DataFrame({'Ventes': stats_ventes, 'Coûts': stats_couts})
stats.round(2)

```
