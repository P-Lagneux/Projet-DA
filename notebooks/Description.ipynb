# Description des données

Le fichier thelook_fr_women_2023_2024.csv contient des données relatives aux ventes de vêtements pour femmes en France sur les années 2023 et 2024. Il est composé de 1 679 lignes et 20 colonnes, qui décrivent les commandes, les articles vendus, les prix, les coûts, les marques, les catégories de produits ainsi que les informations géographiques des clients.

Chaque ligne correspond à un article associé à une commande. Les principales variables disponibles comprennent la date de création de l’article, son prix de vente, son coût, sa marque, sa catégorie et la ville associée à la commande. Le fichier contient également des informations sur le statut des commandes, leur expédition et leur livraison.

```python
import pandas as pd
```

```python
df = pd.read_csv("../data/thelook_fr_women_2023_2024.csv")
print(f"Dimensions : {df.shape[0]:,} lignes × {df.shape[1]} colonnes")

df.info()
df.head()
```
